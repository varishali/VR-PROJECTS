```python
import pandas as pd
from sqlalchemy import create_engine, text
from urllib.parse import quote_plus
import getpass

df = pd.read_csv(
    "data/super_store_clean.csv",
    parse_dates=["order_date", "ship_date"],
    dtype={"postal_code": str}
)
print("CSV rows:", len(df))

password = quote_plus(getpass.getpass("MySQL password: "))

# Pehle database banao (database ke bina connect nahi hota)
server_engine = create_engine(f"mysql+pymysql://root:{password}@localhost:3306")
with server_engine.connect() as conn:
    conn.execute(text("CREATE DATABASE IF NOT EXISTS superstore_db"))
    conn.commit()

# Ab database se connect karke staging table bharo
engine = create_engine(f"mysql+pymysql://root:{password}@localhost:3306/superstore_db")
df.to_sql("stg_superstore", engine, if_exists="replace", index=False, chunksize=1000)

print(pd.read_sql("SELECT COUNT(*) AS staging_rows FROM stg_superstore", engine))
```
