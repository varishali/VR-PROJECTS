## Missing Value Percentage Ke Sath

```python
missing = df.isnull().sum()
missing
```

```python
missing_pct = (missing / len(df) * 100).round(2)
missing_pct
```

```python
pd.DataFrame({"missing":  missing, "percent": missing_pct}).query("missing > 0")
```
