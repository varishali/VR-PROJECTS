## Load aur India Filter

```python
import pandas as pd

df = pd.read_csv("C:/Users/hp/VR-PROJECTS/zomato-restaurant-analysis/data/zomato.csv",encoding="latin-1")
india = df[df['Country Code'] == 1].copy()
india.shape
```

## Cusines ke missing rows

```python
india['Cuisines'].isnull().sum()
india = india.dropna(subset=['Cuisines'])
```

```python
raw = pd.read_csv("C:/Users/hp/VR-PROJECTS/zomato-restaurant-analysis/data/zomato.csv",encoding="latin-1")
print(raw['Has Table booking'].unique())
print(raw['Has Online delivery'].unique())
```

## Yes/No ko (0/1)

```python
for col in ['Has Table booking','Has Online delivery']:
    india[col] = raw.loc[india.index, col].map({'Yes':1, 'No':0})

print(india['Has Table booking'].value_counts(dropna=False))
print(india['Has Online delivery'].value_counts(dropna=False))    
```

```python
india.columns.tolist()
```

## Unnecessary columns hatao

```python
cols_to_drop = ['Switch to order menu', 'Rating color','Country Code', 'Currency']
india = india.drop(columns=cols_to_drop, errors='ignore')
```

## Cuisines ko tod kar alag table banao

```python
cuisines = india[['Restaurant ID', 'Cuisines']].copy()
cuisines['Cuisines'] = cuisines['Cuisines'].str.split(", ")
cuisines = cuisines.explode('Cuisines')
```
